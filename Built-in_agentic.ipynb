## Step 2: Use a built-in search tool


```python
from langchain_community.tools import DuckDuckGoSearchRun
```

## Step 3: Import the DuckDuckGo search tool


```python
search_tool = DuckDuckGoSearchRun()
results = search_tool.invoke("India's population in 2025")
```

```python
print(results)
```

## Step 4: Print the search results


##  Use the Shell Tool


```python
from langchain_community.tools import ShellTool
```

```python
print("Name:", search_tool.name)
print("Description:", search_tool.description)
print("Arguments:", search_tool.args)
```

There are so many other tools in LangChain for agentic AI. The official built-in tools website is the LangChain Tools documentation: https://python.langchain.com/docs/integrations/tools/
